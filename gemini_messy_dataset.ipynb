```python
import pandas as pd
import numpy as np

# Create messy raw data
raw_data = {
    'Customer_ID': ['C101', 'C102', 'C103', 'c104', 'C105', 'C105', 'C107', 'C108', 'C109', 'C110'],
    'Name': [' Alice Smith ', 'Bob Jones', 'charlie brown', 'David Miller', 'Eva Green', 'Eva Green', '  Frank  ', 'Grace Lee', 'HENRY FORD', 'Isabella'],
    'Age': [25, 30, np.nan, '45', 29, 29, 150, -5, 34, 28],
    'Join_Date': ['2023-01-15', '12/05/2022', '2021/08/30', 'Jan 10, 2023', '2023-01-15', '2023-01-15', None, '2023-04-01', '2022-11-11', '05-12-2021'],
    'Salary': ['$65,000', '72000', '$50,000', '85,000', '$90,000', '$90,000', 'UNK', '$45,000', '120000', 'NaN'],
    'Department': ['IT', 'HR', 'it', 'Finance', 'Marketing', 'Marketing', 'IT', 'hr', 'ENGINEERING', 'Finance'],
    'Is_Active': [True, 'Yes', 'False', 1, 'True', 'True', 0, 'No', 'True', np.nan]
}

df = pd.DataFrame(raw_data)
print(df)
```

```python
sum_df = pd.DataFrame({
    'missing': df.isnull().sum(),
    'unique': df.nunique(),
    'type': df.dtypes
})
sum_df
```

```python
df = df.drop_duplicates()
```

```python
df['Name'] = df['Name'].str.strip()
df
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Age']  = df['Age'].astype(int)
df['Age'] = np.where(df['Age'] < 0, df['Age'].median(), df['Age'])
df['Age'] = np.where(df['Age'] > 100, df['Age'].median(), df['Age'])
df
```

```python
df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
df
```

```python
df['Department'] = np.where(df['Department'].str.lower() == 'it', 'IT', df['Department'])
df['Department'] = np.where(df['Department'].str.lower() == 'hr', 'HR', df['Department'])
df['Department'] = np.where(df['Department'].str.lower() == 'finance', 'Finance', df['Department'])
df['Department'] = np.where(df['Department'].str.lower() == 'marketing', 'Marketing', df['Department'])
df
```

```python
df['Is_Active'] = np.where(df['Is_Active'].str.lower() == 'yes', True, df['Is_Active'])
df['Is_Active'] = np.where(df['Is_Active'].str.lower() == 'no', False, df['Is_Active'])
df['Is_Active'] = df['Is_Active'].fillna(False)
df['Is_Active'] = np.where(df['Is_Active'] == 0 , False, df['Is_Active'])
df['Is_Active'] = np.where(df['Is_Active'] == 1, True, df['Is_Active'])
df
```

```python
df['Customer_ID'] = df['Customer_ID'].str.upper()
df['Name'] = df['Name'].str.title()
df
```

```python
df['Join_Date'] = pd.to_datetime(df['Join_Date'], format='mixed', errors='coerce')
df['Join_Date'] = df['Join_Date'].fillna('2022-01-01')

df
```

I cleaned and standardized a messy customer dataset using Pandas to prepare it for analysis. The raw data suffered from several common data quality issues, including duplicate rows, inconsistent text formatting, mixed date formats, and corrupted numerical values. To fix these problems, I dropped duplicate entries, trimmed extra spaces from text fields, and standardized names and departments to consistent casing.

Next, I converted the monetary and numerical fields into proper numeric types. I removed currency symbols, commas, and invalid text markers like "UNK" from the salary column, while also filtering out impossible age values like negative numbers and extreme high outliers. Finally, I parsed mixed date representations into a uniform ISO date format and mapped varied representations of active statuses (such as "Yes", "No", 1, and 0) into standard boolean values. The resulting dataset is clean, properly typed, and ready for downstream reporting or modeling.
