```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
ROOT_DIR = Path.cwd()
orders = ROOT_DIR/'data'/'orders_task.csv'
customers = ROOT_DIR/'data'/'customers_task.csv'
order_items = ROOT_DIR/'data'/'order_items_task.csv'
products = ROOT_DIR/'data'/'products_task.csv'

df1 = pd.read_csv(orders)
df2 = pd.read_csv(customers)
df3 = pd.read_csv(order_items)
df4 = pd.read_csv(products)

df5 = pd.merge(df1, df2, how='left', on='customer_id')
df6 = pd.merge(df5, df3, how='left', on='order_id')
df7 = pd.merge(df6, df4, how='left', on='product_id')
df7

df = df7[df7['status'] == 'Completed']
df7
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='%Y-%m-%d', errors='coerce')
```

Report total realized revenue, total profit, and the overall profit margin.

```python
df['revenue'] = df['quantity'] * df['price']
df['profit'] = df['revenue'] - (df['quantity'] * df['cost'])
df['profit_marg'] = df['profit'] / df['revenue'] * 100
rev = df['revenue'].sum()
prof = df['profit'].sum()
print(f'Total Revenue: {rev:.2f}')
print(f'Total Profit: {prof:.2f}')
print(f'the overall profit margin is: {prof/rev * 100:.2f}%')
df
```

```python
df.columns = df.columns.str.strip()
```

Rank the product categories by revenue and by profit margin.

```python
category = (
    df.groupby('category')[['revenue', 'profit_marg']].agg({
        'revenue': 'sum',
        'profit_marg': 'mean'
    })
)
category
```

Produce a monthly revenue series and its month-over-month growth.

```python
# Ensure order_date is a datetime object first
df['order_month'] = pd.to_datetime(df['order_date']).dt.to_period('M')

monthly_df = (
    df.groupby('order_month')['revenue']
    .sum()
    .reset_index()
    .rename(columns={'revenue': 'monthly_revenue'})
)

monthly_df['mom_change'] = monthly_df['monthly_revenue'].pct_change() * 100
monthly_df
```

For each country, identify the customer contributing the largest share of that
country's revenue.

```python
# 1. Total revenue per customer within each country
cust_df = df.groupby(['country', 'name'])['revenue'].sum().reset_index()

# 2. Add total country revenue and compute individual share percentage
cust_df['country_total'] = cust_df.groupby('country')['revenue'].transform('sum')
cust_df['revenue_share_%'] = (cust_df['revenue'] / cust_df['country_total'] * 100).round(2)

# 3. Filter for the top-contributing customer per country
top_customers = (
    cust_df.sort_values(['country', 'revenue'], ascending=[True, False])
    .groupby('country')
    .first()
    .reset_index()
)

top_customers[['country', 'name', 'revenue', 'revenue_share_%']]
```

Rank the categories by profit and report each category's cumulative share of total
profit.

```python
# 1. Calculate total profit by category and sort descending
cat_df = (
    df.groupby('category')['profit']
    .sum()
    .reset_index()
    .sort_values(by='profit', ascending=False)
    .reset_index(drop=True)
)

# 2. Add rank
cat_df['rank'] = cat_df.index + 1

# 3. Compute cumulative share of total profit
total_profit = cat_df['profit'].sum()
cat_df['profit_share_%'] = (cat_df['profit'] / total_profit * 100).round(2)
cat_df['cum_share_%'] = cat_df['profit_share_%'].cumsum().round(2)

# Final formatted result
cat_df[['rank', 'category', 'profit', 'profit_share_%', 'cum_share_%']]
```

Identify the single highest-revenue product and the single highest-margin product.

```python
# 1. Aggregate total revenue and total profit by product
product_stats = df.groupby('product_name').agg(
    total_revenue=('revenue', 'sum'),
    total_profit=('profit', 'sum')
)

# 2. Calculate true profit margin percentage
product_stats['profit_margin_%'] = (product_stats['total_profit'] / product_stats['total_revenue']) * 100

# 3. Extract the top product for each metric using idxmax()
top_revenue_product = product_stats.loc[[product_stats['total_revenue'].idxmax()]]
top_margin_product = product_stats.loc[[product_stats['profit_margin_%'].idxmax()]]

print("--- Highest-Revenue Product ---")
print(top_revenue_product[['total_revenue', 'profit_margin_%']])

print("\n--- Highest-Margin Product ---")
print(top_margin_product[['total_revenue', 'profit_margin_%']])
```
