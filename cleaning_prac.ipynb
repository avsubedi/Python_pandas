```python
sum = pd.DataFrame({
    'missing': df.isnull().sum(),
    'types': df.dtypes
})
sum
```

```python
df[' Join Date '] = pd.to_datetime(df[' Join Date '], format='mixed', errors='coerce')
df
```

```python
df['Department'] = df['Department'].str.strip()
df['Department'] = np.where(df['Department'] == "n/a", np.nan, df['Department'])
df['Department'] = np.where(df['Department'] == 'eng', 'Engineering', df['Department'])
df
```

```python
df['salary_usd'] = df['salary_usd'].str.replace('$', '')
df['salary_usd'] = df['salary_usd'].str.replace('USD', '')
df['salary_usd'] = df['salary_usd'].str.replace(',', '')
df
```

```python

df['salary_usd'] = pd.to_numeric(df['salary_usd'], errors='coerce')
df['salary_usd'] = np.where(df['salary_usd'] < 0, df['salary_usd'].mean(), df['salary_usd'])
df['salary_usd'] = df['salary_usd'].round(2)
df = df.dropna(subset=['salary_usd'])
df
```

```python
df.columns = df.columns.str.strip()
```

```python
df['Name'] = df['Name'].str.replace('-', '')
df
```

```python
df['performance_score'] = pd.to_numeric(df['performance_score'], errors='coerce')
df['performance_score'] = np.where(df['performance_score'] < 0, df['performance_score'].mean(), df['performance_score'])
df = df.dropna(subset=['performance_score'])
df
```
