```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
ROOT_DIR = Path.cwd()
orders = ROOT_DIR/'data'/'gemini_orders.csv'
df = pd.read_csv(orders)
df
```

```python
sum = pd.DataFrame({
    'missing': df.isnull().sum(),
    'dtypes': df.dtypes
})
sum
```

```python
df['Purchase_Date'] = pd.to_datetime(df['Purchase_Date'], format='mixed', errors='coerce')
df.columns = df.columns.str.strip()
df = df.drop_duplicates(subset=['Product', 'Quantity'], keep='first')
df['Customer_Name'] = df['Customer_Name'].str.strip()
df
```

```python
valid = ('.com', '.org', '.net')

conditions = [
    df['Email'].str.endswith(valid, na=False),
    df['Email'].str.endswith('.co', na=False)
]

choices = [
    df['Email'],
    df['Email'] + 'm'
]

df['Email'] = np.select(conditions, choices, default=df['Email']+'.com')

df
```

```python

df = df[df['Status'] != 'Cancelled']
df
```

```python
df['Price_Per_Unit'] = pd.to_numeric(df['Price_Per_Unit'], errors='coerce')
```

```python
df['Total_Spent'] = df['Quantity'] * df['Price_Per_Unit']
df
```

```python
df = df.dropna(subset=['Quantity', 'Price_Per_Unit'])
df['Quantity'] = np.where(df['Quantity'] < 0, 0, df['Quantity'])
df
```
